```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("sample(1000_data).csv")
data
```

```python
plt.scatter(data.age,data.job)
```

```python
x = data[["age"]]
y = data[["job"]]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size = 0.2)
```

```python
x_train
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
# LogisticRegression yana class eke LogisticRegression() constructor eken model lesa object ekak hada ganii
model = LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

```python
model.predict(x_test)
```

```python
x_test
```

```python
y_test
```

```python
# pahatha paridi apita acuracy eka ganna puluwan(0 th 1 th athara agayak labee)
model.score(x_test,y_test)
```
