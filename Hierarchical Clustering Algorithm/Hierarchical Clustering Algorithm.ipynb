```python
import pandas as pd
data = pd.DataFrame(data = {'x': [0,1.1,1,2,2,4,5,5],
                           'y' : [0,1.5,4,2,3,1,0,4]})
data
```

```python
import matplotlib.pyplot as plt
plt.scatter(data.x,data.y)

#data points name kiriima
numbers = [0,1,2,3,4,5,6,7]
for index,value in enumerate(numbers): plt.annotate(value,(data.x[index], data.y[index]),size =15)
```

```python
#Hierarchical chart eka display kara ganiima

import scipy.cluster.hierarchy as sc
dendrogram = sc.dendrogram(sc.linkage(data,method = 'ward'))
plt.title('Dendrogram')
plt.show()
```

```python
#awashya clusters gana laba diima pahatha paridi kala haka

from sklearn.cluster import AgglomerativeClustering
model = AgglomerativeClustering(n_clusters = 2 , linkage = 'ward')
pred = model.fit_predict(data)
pred
```

```python
data['cluster'] = pred
data
```

```python
cluster1 = data[data['cluster'] == 0]
cluster2 = data[data['cluster']== 1]

plt.scatter(cluster1.x, cluster1.y)
plt.scatter(cluster2.x,cluster2.y)
```
