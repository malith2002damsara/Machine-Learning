```python
x = data.iloc[:,1:5]
# ithatha eke adahasa => iloc = "index location" , : => okkoma rows tika ganna  ,  1:5 => 1  ha 5 wenakm columns tika ganna(1,2,3,4 columns tika)
x
```

```python
y = data.iloc[:,-1]
# -1 yanu anthima column ekai
y
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
# x values tika feature scaling kala yuthuya

x = scaler.fit_transform(x)
x
```

```python
 from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier(n_neighbors = 1)
model.fit(x_train,y_train)
# n_neighbors = 1 yanu 'k' value ekai
```

```python
pred = model.predict(x_test)
```

```python
pred
```

```python
y_test
```

```python
# accuracy eka check kiriima 
from sklearn.metrics import accuracy_score
accuracy =  accuracy_score(y_test,pred)
accuracy
```

```python
# accuracy eka confution matrix eken check kiriima

from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test,pred)
cm
```

```python
# result eka display kara ganiima
# data tika columns widiyata one nisa transpose karai

result = pd.DataFrame(data = [y_test.values,pred], index = ['y_test','pred'])
result.transpose()
```

```python
# hari giya values tika
correct_sum = []
for i in range(1,20):
    model = KNeighborsClassifier(n_neighbors = i)
    model.fit(x_train,y_train)
    pred = model.predict(x_test)
    correct =np.sum(pred == y_test)
    correct_sum.append(correct)
    
```

```python
correct_sum
```

```python
result = pd.DataFrame(data == correct_sum)
result.index = result.index + 1
result.T
```
