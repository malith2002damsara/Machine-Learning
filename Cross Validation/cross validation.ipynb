```python
import pandas as pd
data = pd.read_csv('iris.csv') 
data = data.drop('Id',axis = 1)
data
```

```python
x = data.drop('Species',axis = 1)
x.head(5)
data.head(5)
```

```python
y = data['Species']
y.head(5)
```

```python
#cross val import kiriima

from sklearn.model_selection import cross_val_score
```

```python
#K-Nearest Neighbors

from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier()
cross_val_score(knn,x,y,cv = 5).mean()

# ihatha cv = 5 vyannen wenne data tika piliwelata kotas 5 kata kadagena eewa wena wenama test and train lesa gena accuracy values 5 labee.
# pasuma .mean() magin ema labena accuracy values 5 he avaerage eka ganii
```

```python
#Random Forest

from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier()
cross_val_score(rf,x,y,cv = 5).mean()
```

```python
#Naive Bayes

from sklearn.naive_bayes import GaussianNB
nb = GaussianNB()
cross_val_score(nb,x,y,cv = 5).mean()
```

```python
#Support Vector Machine

from sklearn.svm import SVC
svm = SVC()
cross_val_score(svm,x,y,cv = 5).mean()
```
