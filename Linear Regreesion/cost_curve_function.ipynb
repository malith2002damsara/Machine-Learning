```python
import numpy as np
import matplotlib.pyplot as plt

# number of data points
n = 5

# simple dataset
x = np.array([1, 2, 3, 4, 5])
y = np.array([5, 8, 11, 14, 17])

# start with m=0 and c=0
m = 0
c = 0
learning_rate = 0.01

m_values = []
cost_values = []

for i in range(1, 100):
    y_predict = m * x + c
    # calculate cost
    cost = (1/n) * sum((y - y_predict) ** 2)

    # store values for plotting later
    m_values.append(m)
    cost_values.append(cost)

    # calculate gradients
    dm = -(2/n) * sum(x * (y - y_predict))
    dc = -(2/n) * sum(y - y_predict)

    # update parameters
    m = m - learning_rate * dm
    c = c - learning_rate * dc

    print("m {:.4f}, c {:.4f}, cost {:.4f}, iteration {}".format(m, c, cost, i))

# plot m vs cost
plt.figure(figsize=(8, 5))
plt.scatter(m_values, cost_values, color='blue', s=10)
plt.title("Gradient Descent Progress (m vs cost)")
plt.xlabel("Slope (m)")
plt.ylabel("Cost")
plt.grid(True)
plt.show()
```
